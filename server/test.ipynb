```python
import requests

ticker = "NVDA"
url = f"https://query2.finance.yahoo.com/v8/finance/chart/{ticker}?period1=1699549200&period2=1731319200&interval=1d"
response = requests.get(url, headers={"User-Agent": "Mozilla/5.0"})
data = response.json()

# Parse and print the date and close prices
timestamps = data['chart']['result'][0]['timestamp']
prices = data['chart']['result'][0]['indicators']['quote'][0]['close']

for time, price in zip(timestamps, prices):
    print(f"Date: {time}, Close Price: {price}")
```

```python
import feedparser

RSS_URL = "https://feeds.finance.yahoo.com/rss/2.0/headline?s="
# GENERAL = https://feeds.finance.yahoo.com/rss/2.0/headline?s=^DJI


def fetch_yahoo_finance_news(ticker: str):
    """
    Fetches news articles for a given ticker from Yahoo Finance RSS feed.
    """
    feed = feedparser.parse(RSS_URL + ticker)
    return feed.entries
```

```python
e = fetch_yahoo_finance_news("TSLA")
```
