```python
import requests

ticker = "NVDA"
url = f"https://query2.finance.yahoo.com/v8/finance/chart/{ticker}?period1=1699549200&period2=1731319200&interval=1d"
response = requests.get(url, headers={"User-Agent": "Mozilla/5.0"})
data = response.json()

# Parse and print the date and close prices
timestamps = data['chart']['result'][0]['timestamp']
prices = data['chart']['result'][0]['indicators']['quote'][0]['close']

for time, price in zip(timestamps, prices):
    print(f"Date: {time}, Close Price: {price}")
```

```python
import pandas as pd
df = pd.read_excel("ticker_symbols.xlsx", sheet_name="Stock", skiprows=3, usecols="A:E")
```

```python
def search_dataframe(df, search_term, columns):
    # Create a boolean mask by checking if search_term is in any of the specified columns
    mask = df[columns].apply(lambda x: x.str.contains(search_term, case=False, na=False))
    
    # Get the top 5 matches
    result = df[mask.any(axis=1)].head(5)
    
    return result

# Example usage
search_term = "Apple"  # Modify this to your search term
columns_to_search = ['Name', 'Ticker']  # Specify which columns to search

top_results = search_dataframe(df, search_term, columns_to_search)
top_results.to_json(orient='records', indent=4)
```
